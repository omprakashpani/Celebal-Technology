```python

from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
spark = SparkSession.builder.appName("Spark Basics Assignment").getOrCreate()
```

```python
df = spark.read.csv(
    "../data/Sample - Superstore.csv",
    header=True,
    inferSchema=True
)

df.show(5)
```

```python
df.printSchema()
```

```python
df.columns
```

```python
df = df.dropDuplicates()
```

```python
from pyspark.sql.functions import col

df.filter(col("Sales").isNull()).show()
df.filter(col("Postal Code").isNull()).count()
```

```python

df = df.fillna("Unknown")
df = df.fillna(0)
df.show(5)
```

```python
from pyspark.sql.functions import col

df = df.withColumn("Sales", col("Sales").cast("double"))
```

```python
df.printSchema()
```
