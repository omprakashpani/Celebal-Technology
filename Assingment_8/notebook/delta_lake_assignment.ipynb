```python
incremental_data = [
    (2174, 52000, "Graduation", "Married"),
    (4141, 80000, "PhD", "Together"),
    (5324, 65000, "Master", "Married"),
    (7446, 70000, "Master", "Single"),
    (6177, 40000, "Graduation", "Married"),
    (9001, 45000, "Graduation", "Single"),
    (9002, 60000, "PhD", "Married"),
    (9003, 72000, "Master", "Together"),
    (9004, 55000, "Graduation", "Divorced"),
    (9005, 90000, "PhD", "Single")
]

increment_df = spark.createDataFrame(
    incremental_data,
    ["ID", "Income", "Education", "Marital_Status"]
)

display(increment_df)
```

```python
delta_df.createOrReplaceTempView("cm")

increment_df.createOrReplaceTempView("ci")
```

```python
%sql
merge into cm as target
using ci as source
on target.ID = source.ID
when matched then 
update set
    target.income=source.income,
    target.education=source.education,
    target.marital_status=source.marital_status
    
when not matched then 
insert(
    ID,
    income,
    education,
    marital_status
) values (
    source.ID,
    source.income,
    source.education,
    source.marital_status
);
```
